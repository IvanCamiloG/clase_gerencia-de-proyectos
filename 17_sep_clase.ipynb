La imagen resume un proceso formal para gestionar cambios, variaciones y controlar el cronograma de un proyecto, basado en conceptos del PMBOK 8.ª edición.

Resumen completo
Cambio inesperado: durante el proyecto pueden aparecer situaciones que alteren lo planificado.

Monitoreo y control: se registran observaciones y mediciones para detectar desviaciones respecto al plan.

Identificación de la varianza: se determina qué está cambiando y si requiere una acción. Por ejemplo, la imagen muestra un proyecto con 60 % de trabajo completado, 25 % restante y 15 % de desviación.

Evaluación: se analiza el impacto del cambio en aspectos como costo y tiempo y se documenta la necesidad de actuar.

Comité de Cambios (CCB): grupo encargado de evaluar y aprobar o rechazar cambios solicitados.

Presentación: la solicitud de cambio se lleva formalmente al CCB para su revisión.

Decisión: el comité puede aprobar, aplazar o rechazar la solicitud.

Entregables clave: una vez tomada la decisión, el cambio debe quedar registrado y gestionarse según corresponda.

En síntesis, el proceso busca evitar cambios improvisados: detectar la desviación → evaluar sus efectos → presentar el cambio → obtener una decisión formal → controlar su ejecución.
