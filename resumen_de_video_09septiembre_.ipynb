El material explica los elementos fundamentales para iniciar y evaluar un proyecto. Primero se identifica una necesidad u oportunidad empresarial, es decir, un problema que debe resolverse o una posibilidad que la organización puede aprovechar.

Después se establecen los objetivos del proyecto, procurando que sean claros y medibles. Estos objetivos permiten determinar qué se quiere lograr y cómo se podrá comprobar si el proyecto tuvo éxito.

También se definen los criterios de éxito, que sirven para evaluar los resultados obtenidos. Un aspecto importante es el valor generado por cada unidad de inversión, ya que permite determinar si los recursos utilizados realmente producen beneficios para la organización.

Para comprobar si el proyecto es conveniente, se desarrolla un Caso de Negocio, que analiza su viabilidad económica y los beneficios esperados. Finalmente, se utiliza el Acta Constitutiva, que es el documento que autoriza formalmente el proyecto y otorga la autoridad necesaria para utilizar recursos.

En conjunto, el proceso busca que un proyecto responda a una necesidad real, tenga objetivos definidos, sea económicamente viable y genere valor, antes de avanzar con su ejecución.
