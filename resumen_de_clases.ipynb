La gobernanza de proyectos es el conjunto de principios, reglas, procesos y responsabilidades que permiten dirigir, controlar y tomar decisiones dentro de un proyecto. Su objetivo principal es garantizar que el proyecto avance de acuerdo con los objetivos estratégicos de la organización, manteniendo claridad sobre quién puede tomar decisiones, quién es responsable y cuáles son los límites que debe respetar el equipo.

En los proyectos ágiles, la gobernanza busca encontrar un equilibrio entre control y autonomía. Aunque existen políticas, límites y responsabilidades definidos, los equipos ágiles tienen libertad para organizar su trabajo y tomar decisiones locales de manera rápida.

1. Marco de gobernanza

El marco de gobernanza establece las bases para gestionar el proyecto. Define las reglas, políticas y mecanismos que deben seguirse para garantizar que las decisiones y actividades estén alineadas con los objetivos de la organización.

2. Roles clave del equipo

Cada integrante tiene responsabilidades específicas:

Patrocinador: proporciona recursos, apoyo y dirección estratégica al proyecto.
Product Owner: administra y prioriza el backlog, definiendo qué funcionalidades tienen mayor valor para el usuario.
Scrum Master: ayuda al equipo a trabajar correctamente bajo principios ágiles y elimina obstáculos que puedan afectar el desarrollo.
Desarrolladores: se encargan de construir y entregar el producto o las funcionalidades.

Esta distribución permite que exista claridad sobre las responsabilidades y evita confusiones en la toma de decisiones.

3. Toma de decisiones

En un entorno ágil se busca una autoridad distribuida. Esto significa que no todas las decisiones deben depender de un único jefe o nivel superior.

Los equipos pueden tomar decisiones locales relacionadas con su trabajo, siempre respetando las políticas y límites establecidos por la organización. Esto permite responder más rápidamente a problemas y cambios.

4. Autogobierno ágil

Los equipos ágiles tienen la capacidad de autoorganizarse. Los integrantes colaboran, distribuyen las tareas y buscan soluciones sin depender constantemente de una autoridad externa.

Las reuniones diarias, como el Daily Scrum, ayudan a conocer el avance, identificar obstáculos y coordinar las actividades del equipo.

5. Plan de gobernanza

El plan de gobernanza determina cómo se va a dirigir y controlar el proyecto. Establece aspectos como:

Quién toma determinadas decisiones.
Qué responsabilidades tiene cada rol.
Qué reglas debe seguir el equipo.
Cómo se controlan los riesgos.
Cuándo es necesario involucrar a niveles superiores.
Cómo se realiza el seguimiento del proyecto.
6. Matriz RACI

La Matriz RACI es una herramienta utilizada para definir claramente la participación de las personas en cada actividad del proyecto.

RACI representa:

R – Responsible: persona que realiza la tarea.
A – Accountable: persona que tiene la responsabilidad final de que la tarea se complete correctamente.
C – Consulted: personas que deben ser consultadas antes de tomar una decisión.
I – Informed: personas que deben mantenerse informadas.

Esto ayuda a evitar problemas como tareas sin responsable o varias personas tomando decisiones sobre lo mismo.

7. Políticas

Las políticas establecen las reglas y estándares que deben cumplirse durante el proyecto. Sirven para garantizar que el trabajo se realice de acuerdo con los requisitos de la organización y las normas establecidas.

8. Límites

Los límites establecen hasta dónde puede actuar el equipo de manera autónoma. Pueden estar relacionados con:

Presupuesto.
Seguridad.
Calidad.
Riesgos.
Tecnología.
Cumplimiento de normas.

Estos límites permiten que el equipo tenga libertad, pero sin poner en peligro los objetivos del proyecto.

9. Escalamiento

El escalamiento ocurre cuando un problema o decisión supera la autoridad o los límites del equipo. En esos casos, puede ser necesario involucrar a la PMO, al patrocinador u otros responsables de mayor nivel.

Por ejemplo, si el equipo necesita aumentar considerablemente el presupuesto, puede tener que escalar la decisión al patrocinador.

Conclusión

La gobernanza ágil permite que un proyecto tenga una dirección clara sin eliminar la flexibilidad característica de Agile. Define roles, responsabilidades, políticas, límites y mecanismos de decisión, mientras permite que los equipos se autoorganicen y resuelvan problemas de manera autónoma.

En pocas palabras, la gobernanza busca que cada persona sepa qué debe hacer, quién puede tomar decisiones, cuáles son los límites y cuándo es necesario escalar un problema. Esto ayuda a mejorar el control, la transparencia, la eficiencia y las posibilidades de éxito del proyecto.

## <strong>Introducción</strong>
</div>

El documento que estás leyendo no es una página web estática, sino un entorno interactivo denominado <strong>notebook de Colab</strong>, que permite escribir y ejecutar código.

Por ejemplo, esta es una <strong>celda de código</strong> con una secuencia de comandos Python corta que calcula un valor, lo almacena en una variable y devuelve el resultado:

```
seconds_in_a_day = 24 * 60 * 60
seconds_in_a_day
```

A fin de ejecutar el código en la celda anterior, haz clic en él para seleccionarlo y luego presiona el botón de reproducción ubicado a la izquierda del código o usa la combinación de teclas "Command/Ctrl + Intro". Para editar el código, solo haz clic en la celda y comienza a editar.

Las variables que defines en una celda pueden usarse en otras:

```
seconds_in_a_week = 7 * seconds_in_a_day
seconds_in_a_week
```

Los notebooks de Colab te permiten combinar <strong>código ejecutable</strong> y <strong>texto enriquecido</strong> en un único documento, junto con <strong>imágenes</strong>, <strong>HTML</strong>, <strong>LaTeX</strong> y mucho más. Los notebooks que crees en Colab se almacenan en tu cuenta de Google Drive. Puedes compartir fácilmente los notebooks de Colab con amigos o compañeros de trabajo para que realicen comentarios o los editen. Si quieres obtener más información, consulta la <a href="/notebooks/basic_features_overview.ipynb">Descripción general de Colab</a>. Los notebooks de Colab son notebooks de Jupyter que aloja Colab. Para obtener más información sobre el proyecto Jupyter, visita <a href="https://www.jupyter.org">jupyter.org</a>.

## Ciencia de datos
</div>

Con Colab, puedes aprovechar por completo las bibliotecas más populares de Python para analizar y visualizar datos. La celda de código que se incluye a continuación usa <strong>NumPy</strong> para generar algunos datos aleatorios y <strong>matplotlib</strong> para visualizarlos. Para editar el código, haz clic en la celda y comienza a editar.

Puedes importar datos propios a notebooks de Colab desde tu cuenta de Google Drive &#40;incluso desde hojas de cálculos&#41;, GitHub y muchas otras fuentes. Para obtener más información acerca de la importación de datos y cómo puede usarse Colab para fines relacionados con la ciencia de datos, consulta los vínculos de <a href="#working-with-data">Cómo trabajar con datos</a>.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```
