Resumen: Sastrería y Cierre
Las imágenes presentan el tema de Sastrería, entendido como la adaptación del enfoque y de los procesos de un proyecto al entorno y a las necesidades del trabajo.

Se destacan tres ideas principales:

Sastrería: consiste en ajustar la forma de trabajar de acuerdo con las características y necesidades específicas del proyecto.

Cierre: comprende las actividades necesarias para finalizar un proyecto, como finalizar contratos, confirmar la aceptación, reasignar recursos, archivar registros y transferir operaciones.

Retrospectiva: es un taller en el que se analiza el trabajo realizado para encontrar éxitos, errores y oportunidades de mejora.

La retrospectiva se desarrolla en cuatro pasos:

Reflexión: pensar sobre lo realizado durante la iteración.

Identificación: detectar éxitos y fallos.

Documentación: registrar las lecciones aprendidas.

Transferencia: actualizar y compartir los activos o conocimientos para aplicarlos en futuros proyectos.

En conjunto, las imágenes muestran que cerrar un proyecto no solo significa terminar actividades administrativas, sino también aprender de la experiencia para mejorar proyectos futuros.
