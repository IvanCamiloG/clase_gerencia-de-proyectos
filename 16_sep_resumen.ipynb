Las imágenes presentan un proceso para determinar si un proyecto de software será financieramente viable o si puede convertirse en un “agujero negro” de recursos.

El análisis sigue varios pasos:

Valor del proyecto

Se define cuánto valor puede generar el proyecto.

Se analizan los beneficios esperados frente a la inversión necesaria.

CapEx vs. OpEx

CapEx: gastos de inversión inicial, como infraestructura, equipos o desarrollo.

OpEx: gastos operativos necesarios para mantener funcionando el proyecto.

Métricas de viabilidad

Se utilizan indicadores financieros para determinar si el proyecto resulta conveniente.

Entre los principales están el VAN y la TIR.

Modelo de flujo de caja

Se proyectan los ingresos, costos y gastos del proyecto a lo largo del tiempo.

Con estos datos se obtiene el flujo de caja neto.

Entregables clave

Se identifican los resultados concretos que debe producir el proyecto y que permitirán comprobar su avance y valor.

💰 VAN
El Valor Actual Neto (VAN) representa el valor actual de una serie de flujos de caja futuros, considerando una determinada tasa de descuento.

VAN positivo: los flujos proyectados generan valor por encima de la inversión considerada.

VAN negativo: el proyecto no recuperaría el valor requerido bajo los supuestos utilizados.

📈 TIR
La Tasa Interna de Retorno (TIR) es la tasa de interés o rentabilidad a la que el VAN es igual a cero.

Para calcularla se parte de:
Ingresos → Costos → Flujo neto → Tasa de descuento → VAN/TIR

🎯 Idea principal
El mensaje central es que un proyecto de software no debería evaluarse únicamente desde el punto de vista técnico. También es necesario analizar su viabilidad financiera, estimando ingresos, costos, inversión y flujos de caja para determinar si realmente genera valor.
