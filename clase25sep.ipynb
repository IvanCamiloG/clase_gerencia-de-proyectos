Resumen

El Dashboard Ejecutivo busca mostrar de forma clara y visual el estado financiero y de desempeño de un proyecto utilizando indicadores de EVM (Earned Value Management).

El proceso comienza con la construcción de las curvas base del proyecto:

PV (Planned Value): valor planificado del trabajo.
EV (Earned Value): valor del trabajo realmente ejecutado.
AC (Actual Cost): costo real del trabajo realizado.

A partir de estas variables se analizan las desviaciones del proyecto mediante indicadores como:

CV (Cost Variance): variación de costos.
SV (Schedule Variance): variación del cronograma.
CPI (Cost Performance Index): eficiencia del gasto.
SPI (Schedule Performance Index): eficiencia respecto al cronograma.

El CPI y SPI se comparan con el valor ideal de 1.0. Un valor cercano a 1 indica que el desempeño se encuentra aproximadamente de acuerdo con lo planificado.

Proyecciones del proyecto

El dashboard también permite realizar pronósticos sobre cómo terminará el proyecto:

EAC (Estimate at Completion): costo total estimado al finalizar el proyecto.
ETC (Estimate to Complete): costo estimado que todavía falta para terminarlo.
VAC (Variance at Completion): diferencia esperada entre el presupuesto original y el costo final proyectado.

Por ejemplo, en las imágenes se muestra:

ETC = EAC − AC

Es decir, el costo que todavía se espera gastar corresponde al costo total proyectado menos lo que ya se ha gastado.

TCPÍ

Otro indicador importante es el TCPI (To-Complete Performance Index). Este indica qué nivel de eficiencia debe mantener el equipo desde el momento actual hasta el final del proyecto para alcanzar un objetivo determinado, como el presupuesto establecido.

La presentación utiliza la idea de una propulsión de emergencia: cuando el proyecto presenta un CPI menor a 1, se necesita mejorar la eficiencia para poder recuperar el presupuesto.

En conjunto

El dashboard debe integrar:

Curvas EVM actualizadas al corte.
Indicadores CPI y SPI, comparados con el ideal de 1.0.
Proyecciones financieras, especialmente EAC y VAC.
Meta de eficiencia del equipo mediante TCPI.
Visualizaciones gráficas que permitan identificar rápidamente si el proyecto está dentro del presupuesto y del cronograma.

En pocas palabras: el Dashboard Ejecutivo convierte los datos de EVM en información visual para conocer dónde está el proyecto actualmente, qué tan eficiente está siendo y cuánto se espera que cueste al finalizar.
